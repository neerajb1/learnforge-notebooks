# Day 1 — Build a simple ML model

LearnForge · AI/ML Learning Journey · Day 1, section 4

**Problem:** predict a student's exam score from the number of hours they studied.

**What we will do:** 1. load the data · 2. create and train a linear-regression model · 3. make a prediction · 4. visualise the results.

Run each cell in order with **Shift + Enter**.

```python
# Day 1 — Simple linear regression
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Training data: study hours (X) and exam scores (y)
X = np.array([[1], [2], [3], [4], [5]])
y = np.array([35, 45, 55, 65, 75])

# Create and train the model
model = LinearRegression()
model.fit(X, y)

# Predict the score for 6 study hours
new_hours = np.array([[6]])
prediction = model.predict(new_hours)[0]
print(f"Predicted score for 6 hours: {prediction:.1f}")

# Plot the data, the regression line and the prediction
hours = np.arange(0, 9).reshape(-1, 1)
plt.scatter(X, y, color="#60A5FA", s=80, label="Actual data")
plt.plot(hours, model.predict(hours), color="#F472B6", label="Regression line")
plt.scatter(new_hours, prediction, color="#34D399", s=80, label="Prediction")
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.title("Linear Regression Example")
plt.legend()
plt.grid(True, linestyle="--", alpha=0.3)
plt.show()
```

## Ideas to try

### 1. Change the study hours to 8 and 10

The maximum exam score is 100. What does the model predict?

```python
for h in [8, 10]:
    print(h, "hours ->", round(model.predict([[h]])[0], 1))
```

The model only knows the straight-line pattern in its five examples, not that scores stop at 100. A prediction is an estimate, not a guarantee.

### 2. Add more data points

Add a few more `[hours]` and scores to `X` and `y` above (try some that don't sit on the line), run the first cell again and watch the line move.

### 3. Try a different model: a decision tree

```python
from sklearn.tree import DecisionTreeRegressor

tree = DecisionTreeRegressor(random_state=0)
tree.fit(X, y)
print("Decision tree, 6 hours:", tree.predict([[6]])[0])
```

A decision tree predicts from the examples it has seen, so for 6 hours it answers with the closest example (75), while the straight line continues the pattern (85). Different models make different assumptions.

## 6. Clean and prepare the data

Check the messy copy, then fix one problem at a time, and verify.

```python
# The same student data, as it often arrives: with problems
messy = pd.read_csv("student_data_messy.csv")
print(messy)

# Check missing values
print("\nMissing values:")
print(messy.isnull().sum())

# Check data types
print("\nData types:")
print(messy.dtypes)

# Check for duplicate rows
print("\nDuplicate rows:", messy.duplicated().sum())
```

```python
# Look at the repeated rows first: are they really the same record?
print(messy[messy.duplicated(keep=False)])

# S3 appears twice with identical values, so keep one copy
clean = messy.drop_duplicates().copy()
print("\nRows before:", len(messy), "after:", len(clean))
```

```python
# Attendance was stored as text ("60%"): strip the % sign and convert to numbers
clean["Attendance"] = pd.to_numeric(clean["Attendance"].str.rstrip("%"))

# Passed mixes True/False with yes/no: map every spelling to a real Boolean
clean["Passed"] = clean["Passed"].map({"True": True, "yes": True, "False": False, "no": False})

print(clean.dtypes)
```

```python
# One option: drop the rows with a missing value (shown here, not kept)
dropped = clean.dropna(subset=["StudyHours"])
print("Rows if we drop them:", len(dropped), "of", len(clean))
```

```python
# The option we keep: fill the missing study hours with the median
median_hours = clean["StudyHours"].median()
print("Median study hours:", median_hours)
clean["StudyHours"] = clean["StudyHours"].fillna(median_hours)
print(clean[clean["Student"] == "S2"])
```

```python
# Verify: no missing values, no duplicates, the right types
print(clean)
print("\nMissing values:", clean.isnull().sum().sum())
print("Duplicate rows:", clean.duplicated().sum())
print("\nData types:")
print(clean.dtypes)

# Save the cleaned data
clean.to_csv("student_data_clean.csv", index=False)
```

## 7. Hands-on: the complete workflow

Load the raw data, inspect it, clean it, analyze it and chart it, in one cell.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load the raw data
df = pd.read_csv("student_data_messy.csv")

# Inspect
print("Shape:", df.shape)
print("Missing values:", df.isnull().sum().sum(), "| Duplicate rows:", df.duplicated().sum())

# Clean: the same steps as Section 4
df = df.drop_duplicates().copy()
df["Attendance"] = pd.to_numeric(df["Attendance"].str.rstrip("%"))
df["Passed"] = df["Passed"].map({"True": True, "yes": True, "False": False, "no": False})
df["StudyHours"] = df["StudyHours"].fillna(df["StudyHours"].median())

# Analyze
print("\nShape after cleaning:", df.shape)
print("Correlation with ExamScore:")
print(df[["StudyHours", "Attendance", "PreviousScore"]].corrwith(df["ExamScore"]).round(2))
print("\nStudents scoring 40-80:", df["ExamScore"].between(40, 80).sum(), "of", len(df))

# Visualize: study hours vs exam score
plt.figure(figsize=(6, 4))
plt.scatter(df["StudyHours"], df["ExamScore"])
plt.xlabel("Study hours")
plt.ylabel("Exam score")
plt.title("Study hours vs exam score")
plt.grid(True)
plt.show()
```
